```python
def adstock_transform(spend, decay):
    adstock = np.zeros_like(spend)
    adstock[0] = spend[0]
    for t in range(1, len(spend)):
        adstock[t] = spend[t] + decay * adstock[t-1]
    return adstock
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

def optimal_decay(df, spend_column, target='sales'):
    X_raw = df[[spend_column]].values
    y = df[target].values
    best_decay = None
    best_r2 = -999
    for decay in np.arange(0.1, 0.95, 0.05):
        adstocked = adstock_transform(X_raw.flatten(), decay)
        model = LinearRegression().fit(adstocked.reshape(-1,1), y)
        r2 = model.score(adstocked.reshape(-1,1), y)
        if r2 > best_r2:
            best_r2 = r2
            best_decay = decay
    return best_decay, best_r2
```

```python
channels = ['tv_spend','radio_spend','internet_spend','news_spend','sms_spend']

optimal_decays = {}

for ch in channels:
    decay, score = optimal_decay(df, ch)
    optimal_decays[ch] = decay
    print(f"Optimal decay for {ch}: {decay}   R²: {score:.4f}")
```

```python
for ch in channels:
    decay = optimal_decays[ch]
    df[ch + '_adstock'] = adstock_transform(df[ch].values, decay)
```

```python
# def hill_saturation(adstock, S=0.5, k=1.5):
#     return (adstock ** k) / ((adstock ** k) + (S ** k))
# def hill_saturation(x_norm, S=0.5, k=1.5):
#     return (x_norm ** k) / ((x_norm ** k) + (S ** k))

def hill_saturation_from_adstock(adstock_array, S=0.5, k=2.0):
    x = adstock_array.astype(float)
    x_norm = x / x.max() if x.max() != 0 else x  # avoid division by zero
    numerator = x_norm ** k
    denominator = numerator + (S ** k)
    sat = numerator / denominator

    return sat
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RidgeCV
from sklearn.metrics import r2_score, mean_squared_error

X = df[feature_cols].values
y = df['sales'].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

print("Train shape:", X_train.shape)
print("Test shape :", X_test.shape)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

print("Scaled train shape:", X_train_scaled.shape)
print("Scaled test  shape:", X_test_scaled.shape)

alphas = [0.01, 0.1, 1, 10, 100]
ridge = RidgeCV(alphas=alphas, cv=5)
ridge.fit(X_train_scaled, y_train)

print("Chosen alpha:", ridge.alpha_)

y_train_pred = ridge.predict(X_train_scaled)
y_test_pred  = ridge.predict(X_test_scaled)

print("Train R²:", r2_score(y_train, y_train_pred))
print("Test  R²:", r2_score(y_test, y_test_pred))
print("Test RMSE:", np.sqrt(mean_squared_error(y_test, y_test_pred)))
```

```python
coef_df = pd.DataFrame({
    'feature': feature_cols,
    'coefficient': ridge.coef_
}).sort_values(by='coefficient', ascending=False)

coef_df
```

```python
channel_mapping = {
    'tv_spend': 'tv_spend_sat',
    'radio_spend': 'radio_spend_sat',
    'internet_spend': 'internet_spend_sat',
    'news_spend': 'news_spend_sat',
    'sms_spend': 'sms_spend_sat'
}

for spend_col, sat_col in channel_mapping.items():
    contrib_col = sat_col + '_contribution'

    total_spend = df[spend_col].sum()
    total_contribution = df[contrib_col].sum()

    if total_spend != 0:
        roi = total_contribution / total_spend
    else:
        roi = np.nan

    print(f"{spend_col} → ROI: {roi:.6f}")
```

```python
!pip install pulp

import pulp
import numpy as np
import pandas as pd
```

```python
rois = {
    'tv_spend': 38.131107,
    'radio_spend': 62.918056,
    'internet_spend': 0.782982,
    'news_spend': -3068.519608,
    'sms_spend': -175.393292
}
channels = list(rois.keys())
hist_spend = {ch: df[ch].sum() for ch in channels}
hist_spend
```

```python
total_budget = sum(hist_spend.values())
print("Total historical budget:", total_budget)
```

```python
model = pulp.LpProblem("Marketing_Budget_Optimization", pulp.LpMaximize)
spend_vars = {
    ch: pulp.LpVariable(ch, lowBound=0)
    for ch in channels
}
```

```python
model += pulp.lpSum(rois[ch] * spend_vars[ch] for ch in channels), "Total_Incremental_Sales"
```

```python
model += pulp.lpSum(spend_vars[ch] for ch in channels) == total_budget, "Budget_Constraint"
```

```python
for ch in channels:
    model += spend_vars[ch] <= 1.2 * hist_spend[ch], f"Max_{ch}"
```

```python
model += spend_vars['news_spend'] == 0, "No_Newspaper"
model += spend_vars['sms_spend'] == 0, "No_SMS"
```

```python
model.solve(pulp.PULP_CBC_CMD(msg=False))
print("Status:", pulp.LpStatus[model.status])
```

```python
opt_spend = {ch: spend_vars[ch].varValue for ch in channels}
opt_spend
```

```python
opt_incremental_sales = sum(rois[ch] * opt_spend[ch] for ch in channels)
print("Optimal incremental sales:", opt_incremental_sales)
```

```python
baseline_incremental_sales = sum(rois[ch] * hist_spend[ch] for ch in channels)
print("Baseline incremental sales (historical allocation):", baseline_incremental_sales)

uplift = opt_incremental_sales - baseline_incremental_sales
print("Uplift vs historical allocation:", uplift)
```
